```python
# Librerias

import pandas as pd
import numpy as np
import joblib
```

```python
# Carga Modelo: 

best_model = joblib.load(
    "best_model.pkl"
)

print("Modelo cargado correctamente")
```

```python
# Carga archivo entregado: 

df_pred = pd.read_excel(
    "../data/raw/smoking_prediction_entrega.xlsx"
)

```

```python
print(df_pred.shape)

df_pred.head()
```

```python
# Guardar ID para el archivo final

id_original = df_pred["ID"].copy()
```

```python
# Eliminar ID

df_pred.drop(
    columns=["ID"],
    inplace=True
)
```

```python
#Verificar estructura:

df_pred.columns.tolist()
```

```python
print(df_pred.shape)
print(df_pred.columns.tolist())
```

```python
'BMI' in df_pred.columns
```

```python
'WHR' in df_pred.columns
```

```python
'chol_hdl_ratio' in df_pred.columns
```

```python
'tri_hdl_ratio' in df_pred.columns
```

```python
print(df_pred.shape)
print(df_pred.columns.tolist())
```

```python
df_pred["BMI"] = (
    df_pred["weight(kg)"] /
    ((df_pred["height(cm)"] / 100) ** 2)
)

df_pred["WHR"] = (
    df_pred["waist(cm)"] /
    (df_pred["height(cm)"] / 100)
)

df_pred["chol_hdl_ratio"] = (
    df_pred["Cholesterol"] /
    df_pred["HDL"]
)

df_pred["tri_hdl_ratio"] = (
    df_pred["triglyceride"] /
    df_pred["HDL"]
)
```

```python
print(df_pred.shape)
```

```python
print(df_pred.columns.tolist())
```

### Genera predicciones

```python
# Generar predicciones:

predicciones = best_model.predict(
    df_pred
)

print(predicciones[:20])
```

```python
# Verificar distribución

pd.Series(predicciones).value_counts()
```
